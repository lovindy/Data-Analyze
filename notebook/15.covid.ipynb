```python
import folium  # Import the folium library to create interactive maps
import pandas as pd  # Import pandas for data manipulation and analysis

# Define the base URL for the data files
url = 'https://raw.githubusercontent.com/python-visualization/folium/master/examples/data'

# Specify the URL for the GeoJSON file containing the state boundaries
state_geo = f"{url}/us-states.json"

# Specify the URL for the CSV file containing unemployment data
state_unemployment = f"{url}/US_Unemployment_Oct2012.csv"

# Read the CSV file into a pandas DataFrame
state_data = pd.read_csv(state_unemployment)

# Create a Folium map object centered at the specified latitude and longitude
# with an initial zoom level of 3
m = folium.Map(location=[48, -102], zoom_start=3)

# Add a Choropleth layer to the map
# Choropleth maps are used to represent the distribution of a variable (unemployment rate in this case)
# across different geographic regions (US states in this case)
folium.Choropleth(
    geo_data=state_geo,  # Path to the GeoJSON file containing state boundaries
    name='choropleth',  # Name of the layer (used in the layer control)
    data=state_data,  # Data source for the choropleth map
    columns=["State", "Unemployment"],  # Columns in the data to use: state names and unemployment rates
    key_on="feature.id",  # Key to match the data with the GeoJSON features
    fill_color="YlGn",  # Color scheme for the map (ColorBrewer scheme)
    fill_opacity=0.7,  # Opacity of the filled regions
    line_opacity=0.2,  # Opacity of the boundary lines
    legend_name="Unemployment Rate (%)"  # Legend title
).add_to(m)  # Add the choropleth layer to the map

# Add a layer control to the map, allowing the user to toggle the choropleth layer on and off
folium.LayerControl().add_to(m)

# Display the map (this line is implicit in Jupyter Notebooks and may not be needed in a script)
m
```

```python
import json
import folium
import pandas as pd

# Define the center of the map
center = [35.762887375145795, 84.08313219586536]

# Create a Folium map object
m = folium.Map(
    location=center,
    zoom_start=2,
    max_bounds=True,
    min_zoom=1,
    min_lat=-84,
    max_lat=84,
    min_lon=-175,
    max_lon=187
)

geo_path = "./data/World_Countries_(Generalized).geojson"

# Read the GeoJSON file
with open(geo_path, 'r', encoding='utf-8') as f:
    json_data = json.load(f)

# Check if 'total_vaccinations_per_hundred' column exists in total_df
if 'total_vaccinations_per_hundred' not in total_df.columns:
    raise ValueError("total_vaccinations_per_hundred column not found in total_df")

folium.Choropleth(
    geo_data=json_data,
    data=total_df,
    columns=[total_df.index, 'total_vaccinations_per_hundred'],
    key_on='properties.COUNTRY',
    fill_color='RdYlGn',
    fill_opacity=0.7,
    line_opacity=0.5,
).add_to(m)

folium.LayerControl().add_to(m)

# Display the map
m
```
