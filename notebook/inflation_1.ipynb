```python
import pandas as pd
import seaborn as sns
import datetime
import numpy as np
```

```python
# Sample data
data = {
    'Year': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'CPI': [218.1, 224.9, 229.6, 233.0, 236.7, 237.0, 240.0, 245.1, 251.1, 255.7, 258.8, 271.0, 283.2],
    'PPI': [180.2, 186.1, 191.2, 194.8, 198.2, 200.1, 201.5, 203.2, 207.2, 210.5, 215.3, 222.6, 230.4],
    'Unemployment Rate': [9.6, 9.0, 8.1, 7.4, 6.2, 5.3, 4.9, 4.4, 3.9, 3.7, 8.1, 6.0, 4.2],
    'GDP Growth Rate': [2.6, 1.6, 2.2, 1.8, 2.5, 2.9, 1.6, 2.4, 2.9, 2.3, -3.5, 5.7, 2.1]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
import matplotlib.pyplot as plt

# Calculate moving averages
df['CPI_MA'] = df['CPI'].rolling(window=3).mean()
df['PPI_MA'] = df['PPI'].rolling(window=3).mean()
df['Unemployment_MA'] = df['Unemployment Rate'].rolling(window=3).mean()
df['GDP_Growth_MA'] = df['GDP Growth Rate'].rolling(window=3).mean()

# Plot the data
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.plot(df['Year'], df['CPI'], label='CPI')
plt.plot(df['Year'], df['CPI_MA'], label='CPI Moving Average', linestyle='--')
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(df['Year'], df['PPI'], label='PPI')
plt.plot(df['Year'], df['PPI_MA'], label='PPI Moving Average', linestyle='--')
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(df['Year'], df['Unemployment Rate'], label='Unemployment Rate')
plt.plot(df['Year'], df['Unemployment_MA'], label='Unemployment Moving Average', linestyle='--')
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(df['Year'], df['GDP Growth Rate'], label='GDP Growth Rate')
plt.plot(df['Year'], df['GDP_Growth_MA'], label='GDP Growth Moving Average', linestyle='--')
plt.legend()

plt.tight_layout()
plt.show()
```
