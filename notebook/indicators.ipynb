```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/fredgraph (1).csv")
```

```python
# rename the column name
df.rename(columns={'DATE': 'Date', 'CORESTICKM159SFRBATL': 'CPI', 'PPIACO': 'PPI', 'UNRATE': 'UNEMPLOYMENT'}, inplace=True)
df.head()
```

```python
#  set the date column as the index
df.set_index('Date', inplace=True)
df.head()
```

```python
# Data Visualization: Plot CPI and PPI over time to observe trends and patterns.
plt.figure(figsize=(14, 7))
plt.plot(df.index, df['CPI'], label='Consumer Price Index')
plt.plot(df.index, df['PPI'], label='Producer Price Index')
plt.title('CPI and PPI Over Time')
plt.xlabel('Date')
plt.ylabel('Index Level')
plt.xticks(rotation=90)
plt.legend()
plt.show()
```

```python
# Step 2: Descriptive Statistics and Initial Analysis

# calculate the mean
cpi_mean = df['CPI'].mean()
ppi_mean = df['PPI'].mean()

# calculate the median
cpi_median = df['CPI'].median()
ppi_median = df['PPI'].median()


# calculate the standard deviation
cpi_std = df['CPI'].std()
ppi_std = df['PPI'].std()

# correlation analysis
correlation = df['CPI'].corr(df['PPI'])

```

```python
correlation
```

```python
# create new dataframe to store the CPI and PPI
df_cpi_ppi = pd.DataFrame({'CPI': df['CPI'], 'PPI': df['PPI']})

# plot a scatter plot to visualize the correlation
sns.scatterplot(x='CPI', y='PPI', data=df_cpi_ppi)
plt.show()
```

```python
# display the heatmap leading to more understanding
sns.heatmap(df_cpi_ppi.corr(), annot=True)
plt.show()
```

```python
# display the regplot
sns.regplot(x='CPI', y='PPI', data=df_cpi_ppi)
plt.show()
```
