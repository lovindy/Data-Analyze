```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Define the dataset
data_cars = {
    'Brand': [
        'Toyota Corolla', 'Honda Civic', 'Ford Focus', 'Chevrolet Cruze', 'BMW 3 Series', 'Audi A4', 'Mercedes-Benz C-Class', 
        'Volkswagen Golf', 'Nissan Altima', 'Hyundai Elantra', 'Kia Optima', 'Mazda 3', 'Subaru Impreza', 'Lexus IS', 
        'Acura TLX', 'Infiniti Q50', 'Volvo S60', 'Jaguar XE', 'Cadillac ATS', 'Alfa Romeo Giulia', 'Mitsubishi Lancer', 
        'Buick Verano', 'Chrysler 200', 'Dodge Dart', 'Fiat 500', 'Mini Cooper', 'Tesla Model 3', 'Porsche 911', 
        'Ferrari 488', 'Lamborghini Huracan'
    ],
    'Cars Listings': [
        150, 200, 250, 180, 220, 190, 210, 230, 240, 175, 165, 195, 185, 205, 215, 225, 235, 245, 255, 265, 
        160, 170, 180, 190, 200, 210, 220, 230, 240, 250
    ]
}

# Create DataFrame
df_cars = pd.DataFrame(data_cars)

# Generate a list of RGB colors
colors = np.random.rand(30, 3)  # Generates 30 random RGB colors

# Plot the bar chart
plt.figure(figsize=(12, 6))
plt.bar(x=df_cars['Brand'], height=df_cars['Cars Listings'], color=colors)
plt.xticks(rotation=90)
plt.show()
```

```python
df_weSchool = pd.read_csv("./data/w3school.csv")
df_weSchool
```

```python
df_weSchool['Date'] = pd.to_datetime(df_weSchool['Date'])
# removing the duplicate values
df_weSchool.drop_duplicates(subset=['Pulse'], inplace=True)
# removing the empty cells
df_weSchool.dropna(subset=['Calories'], inplace=True)
# fix wrong data in row 7 from 450 to 45
df_weSchool.loc[7, 'Duration'] = 45
df_weSchool.loc[9, 'Pulse'] = 100
 
# assign the value 101 to the rows where the pulse is less than 100
for x in df_weSchool.index:
    if df_weSchool.loc[x, "Pulse"] < 100:
        df_weSchool.loc[x, "Pulse"] = 101

# assign the value 400 to the rows where the calories is greater than 400
for y in df_weSchool.index:
    if df_weSchool.loc[y, "Calories"] > 400:
        df_weSchool.loc[y, "Calories"] = 400
        
# drop the rows where the calories is greater than 400
for y in df_weSchool.index:
    if df_weSchool.loc[y, "Calories"] < 250:
        df_weSchool.drop(y, inplace=True)
```
