```python
import pandas as pd
import requests
import matplotlib.pyplot as plt

# Fetch Google stock price data from Alpha Vantage
api_key = 'YOUR_API_KEY'
symbol = 'GOOGL'
url = f'https://www.alphavantage.co/query?function=TIME_SERIES_DAILY&symbol={symbol}&apikey={api_key}&outputsize=full&datatype=csv'

# Download the CSV data
response = requests.get(url)
with open('google_stock_prices.csv', 'wb') as file:
    file.write(response.content)

# Load data into a DataFrame
df = pd.read_csv('google_stock_prices.csv')
# set timestamp
df['timestamp'] = pd.to_datetime(df['timestamp'])

# sort index in accending order
df.sort_index(ascending=False, inplace=True)
df.set_index('timestamp', inplace=True)

# Plot closing prices
plt.figure(figsize=(14, 7))
plt.plot(df.index, df['close'], label='Google Stock Price')
plt.title('Google Stock Price Over Time')
plt.xlabel('Date')
plt.ylabel('Close Price (USD)')
plt.legend()
plt.show()
```
