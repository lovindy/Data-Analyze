```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Sample DataFrame
data = {
    'Year': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'CPI': [218.1, 224.9, 229.6, 233.0, 236.7, 237.0, 240.0, 245.1, 251.1, 255.7, 258.8, 271.0, 283.2],
    'PPI': [180.2, 186.1, 191.2, 194.8, 198.2, 200.1, 201.5, 203.2, 207.2, 210.5, 215.3, 222.6, 230.4],
}

df = pd.DataFrame(data)

# Plot CPI and PPI
plt.figure(figsize=(12, 6))
plt.plot(df['Year'], df['CPI'], marker='o', linestyle='-', color='blue', label='CPI')
plt.plot(df['Year'], df['PPI'], marker='o', linestyle='-', color='green', label='PPI')
plt.xlabel('Year')
plt.ylabel('Index Value')
plt.title('CPI and PPI Over Time')
plt.legend()
plt.grid(True)
plt.show()

# Calculate CPI inflation rate
df['CPI Inflation Rate'] = df['CPI'].pct_change() * 100
print(df[['Year', 'CPI', 'CPI Inflation Rate']])

# Check if CPI values are increasing
cpi_increasing = df['CPI'].is_monotonic_increasing
print(f'CPI values are generally increasing: {cpi_increasing}')

# Prepare data for regression
X = df['Year']
y = df['CPI']
X = sm.add_constant(X)  # Add a constant term to the model
model = sm.OLS(y, X).fit()
print(model.summary())

# Calculate CPI growth rate
df['CPI Growth Rate'] = df['CPI'].pct_change() * 100
print(df[['Year', 'CPI', 'CPI Growth Rate']])

# Calculate correlation between CPI and PPI
correlation = df['CPI'].corr(df['PPI'])
print(f'Correlation between CPI and PPI: {correlation}')

# Identify years with high CPI inflation
high_inflation_years = df[df['CPI Inflation Rate'] > 2.0]  # Example threshold
print("Years with high CPI inflation:\n", high_inflation_years)
```

```python
import pandas as pd

# Sample DataFrame for PPI
data = {
    'Year': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'PPI': [180.2, 186.1, 191.2, 194.8, 198.2, 200.1, 201.5, 203.2, 207.2, 210.5, 215.3, 222.6, 230.4]
}

df = pd.DataFrame(data)

# Calculate PPI inflation rate
df['PPI Inflation Rate'] = df['PPI'].pct_change() * 100

# Print PPI and PPI Inflation Rate
print(df[['Year', 'PPI', 'PPI Inflation Rate']])

# Plot PPI
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(df['Year'], df['PPI'], marker='o', linestyle='-', color='green', label='PPI')
plt.xlabel('Year')
plt.ylabel('PPI Index Value')
plt.title('Producer Price Index (PPI) Over Time')
plt.legend()
plt.grid(True)
plt.show()

# Example Output of PPI Data
print(df)
```
